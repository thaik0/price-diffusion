# Stage 13 — Mechanism Regression

**Confirmatory hypothesis:** after controlling for event characteristics, equipment status, standardized shock magnitude, and prior semiconductor momentum may explain the economic magnitude of post-event peer CAR, convergence, and initiator reversal. This tests the mechanisms motivated in Stage 12B; it does not search across new subgroups or establish causality.

## Methodology

The reusable source module estimates separate 1-, 5-, and 10-day OLS models with HC3 standard errors. Predictors are the pre-specified equipment indicator, absolute standardized shock, strictly lagged one-month semiconductor momentum, strictly pre-event abnormal volume, and event direction. A single targeted quadratic-shock extension formally tests the Stage 12B nonlinearity finding. `EconomicPeerIndicator` is retained in the event characteristics but is constant in this frozen baseline economic-peer sample, so it is not identified in the regression. Coefficients, standard errors, 95% confidence intervals, sample sizes, and R-squared are reported.

## Interpretation and limitations

Economic magnitude and interval width are primary; p-values are included only as context. The models are associational, have low explanatory power, omit broad-market volatility because the frozen baseline has no market-return series, and condition on detected events. Stage 12B supplied the hypotheses, while these fixed specifications are Stage 13 confirmatory tests.
