## Interpretation and limitations

Economic peer catch-up averaged +0.34% at five days, compared with −0.27% for correlation peers and +0.01% for the broad basket. The within-event economic-minus-correlation difference is more informative than any single level because it holds the initiator and event conditions fixed.

The paired design does not prove that economic similarity causes diffusion. Correlation portfolios may capture different exposures, international closes are non-synchronous, and the common sample falls from 208 events at one day to 62 at ten days. Stage 12B was exploratory; the definitions, lookback, and paired contrast are fixed here.
