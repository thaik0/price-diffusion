# Stage 13 — Market Regime Analysis

**Confirmatory hypothesis:** lower prior semiconductor-sector returns are associated with stronger subsequent convergence after controlling for event characteristics. Both prior one-month and three-month momentum are tested.

## Methodology

Continuous HC3 regressions are primary. Momentum is the strictly lagged compounded semiconductor-factor return over 21 or 63 trading days; sector volatility is the strictly lagged 21-day annualized volatility. Sign-based weak/nonpositive versus strong/positive comparisons use a zero-return economic threshold rather than optimized quantiles. Models report confidence intervals, sample sizes, and R-squared. Broad-market volatility is explicitly unavailable because no broad-market return series exists in the frozen baseline.

## Interpretation and limitations

Continuous coefficients are the main mechanism evidence; the sign split is descriptive. Regimes are measured from the same semiconductor panel, broad-market volatility is unavailable, and event selection plus overlapping windows limit inference. These confirmatory specifications test the prior exploratory pattern and should not be interpreted as a timing strategy or causal effect.
