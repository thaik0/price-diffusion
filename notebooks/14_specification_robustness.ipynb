# Stage 14 — Specification Robustness

## Research question

**Does the economic peer advantage survive reasonable methodological choices?**

Stage 13 estimates depend on a frozen 3-sigma/5% event rule, selected horizons, semiconductor-adjusted returns, equal-weight peers, and the core universe. This notebook varies one economically justified choice at a time. It does not search for a preferred combination, optimize thresholds, or remove weak results.

## Pre-specified tests and safeguards

- Event thresholds: 2, 2.5, 3, and 4 trailing sigma; the 5% floor and every other detection rule remain fixed.
- Horizons: 1, 3, 5, 10, and 20 trading days.
- Returns: raw and semiconductor-factor adjusted. A genuine market-adjusted model is reported unavailable because the frozen data contain no broad-market benchmark.
- Peer weights: equal weights. Market-cap weights are reported unavailable because market capitalization is missing.
- Universe: core, the pre-existing point-in-time core-plus-extension flag, and exclusion of questionable firms. The latter equals core because all 54 reviewed firms are marked CLEAR.
- Trailing-correlation peers use only pre-event information and are frozen on the event date.

## Interpretation

The paired economic-minus-correlation advantage is positive and its bootstrap interval is above zero at 1, 3, and 5 days. It becomes imprecise at 10 days and is uninformative at 20 days because only two events are complete under all three peer definitions. Absolute five-day economic-peer catch-up stays positive across the nearby event thresholds but its conventional intervals include zero; total convergence declines as the cutoff rises. Core-plus-extension raises total convergence but leaves peer catch-up close to the core estimate, so that universe result should not be read as stronger diffusion.

The strongest claim therefore survives only in a narrow form: economic peers outperform the fixed correlation-peer alternative over short and medium horizons. Positive absolute catch-up and persistent total convergence remain weaker claims.
