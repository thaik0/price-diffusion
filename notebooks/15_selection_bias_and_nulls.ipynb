# Stage 15 — Selection Bias and Null Models

## Research question

**Could the observed economic-peer effect arise from the event-selection process itself?**

Extreme company-minus-peer moves are selected after they occur, so mechanical regression to the mean can resemble convergence. The primary null preserves contemporaneous correlation and volatility regimes, destroys the original lead-lag ordering, redetects events, and reruns the event study. Random-peer and matched pseudo-event placebos test peer identity and event timing separately.

## Fixed production design

The stored production run uses 200 random-peer iterations, 200 pseudo-event iterations, and 500 selection-preserving simulations with seed 42. Pseudo-events retain the initiator and exactly match subsector, strictly trailing volatility regime, lagged semiconductor-sector condition, and trailing liquidity bucket; dates within ten calendar days of any real event for that firm are excluded. The sector condition is a disclosed proxy because no broad-market benchmark exists.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display
from price_diffusion.paths import PROJECT_ROOT

output_dir = PROJECT_ROOT / "outputs" / "robustness" / "nulls"
placebo = pd.read_csv(output_dir / "placebo_results.csv")
simulation = pd.read_csv(output_dir / "simulation_distribution.csv")
null_comparison = pd.read_csv(output_dir / "observed_vs_null_results.csv")
validation = pd.read_csv(output_dir / "validation.csv")
```

```python
five_day = placebo.loc[
    placebo["horizon"].eq(5)
    & placebo["outcome"].isin(["peer_catchup", "initiator_reversal", "convergence"]),
    ["method", "outcome", "mean", "ci_lower", "ci_upper", "sample_size", "experiment_iterations"],
]
display(five_day)
```

```python
display(null_comparison.loc[
    null_comparison["horizon"].eq(5)
    & null_comparison["outcome"].isin(["peer_catchup", "convergence"])
])
display(Image(filename=str(output_dir / "observed_vs_null_distribution.png")))
```

```python
event_counts = simulation[["null_iteration", "detected_event_count"]].drop_duplicates()
display(event_counts["detected_event_count"].describe())
display(validation)
assert validation["passed"].all()
```

## Interpretation

At five days, observed peer catch-up is +0.22%, versus approximately zero for random peers and matched pseudo-events. That directional contrast favors economically selected peer identity and true event timing, but the placebo intervals overlap the observed estimate.

The stricter selection-preserving null has mean five-day peer catch-up near zero. The observed statistic is at the 83rd null percentile with an upper-tail empirical p-value of about 0.17. Five-day convergence is at the 87th percentile with p about 0.13. The observed outcomes are therefore larger than the average null draw but are not unusually extreme under this selection-aware benchmark. Selection bias is not sufficient to erase the point estimates, but it materially weakens confidence that they represent a persistent diffusion mechanism.
