Event, firm, and five-date-block bootstraps preserve different dependence structures rather than assuming every event is independent.

## Interpretation

Five-day peer catch-up is near zero early, negative in 2019–2021, and positive in 2022–2025; every period interval includes zero. The result is therefore not temporally uniform and is strongest in the recent subsample. No single initiator dominates by event share, but the 173 valid five-day observations reduce to 20 firm clusters and 138 unique event dates. About 88% of events have overlapping post-event windows, 35% occur on dates with multiple sample events, and repeated firms are pervasive.

Event, firm, and date-block bootstrap intervals for absolute five-day peer catch-up all include zero. The similar point estimates across methods are reassuring, but dependence-aware uncertainty confirms that absolute catch-up remains imprecise.
