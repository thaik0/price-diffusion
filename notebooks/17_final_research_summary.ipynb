## Recommended interpretation

The result I would defend is narrow: economic relationships identified more short-horizon peer adjustment than the fixed trailing-correlation rule. I would not turn that into a broad claim of persistent price discovery. Absolute catch-up is imprecise under dependence-aware bootstraps, the time split is uneven, equipment convergence is offset by initiator continuation, and the selection-preserving null is not rejected at conventional levels. The evidence is consistent with limited, episodic peer adjustment—not a universal or causal diffusion mechanism.
